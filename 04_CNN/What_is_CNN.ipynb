# What is CNN

**CNN** stands for **Convolutional Neural Network**, a type of deep learning model primarily used for image processing and computer vision tasks.

## Key Characteristics

- **Convolutional Layers**: Apply filters to input data to detect features like edges, textures, and patterns
- **Pooling Layers**: Reduce spatial dimensions and extract dominant features
- **Fully Connected Layers**: Perform classification based on extracted features

## Common Applications

- Image classification
- Object detection
- Facial recognition
- Medical imaging
- Autonomous vehicles

## Architecture

CNNs typically follow this structure:
1. Input layer
2. Convolutional layers (feature extraction)
3. Pooling layers (dimensionality reduction)
4. Flattening layer
5. Fully connected layers (classification)

## Advantages

- Automatically learns features from data
- Efficient for processing images due to weight sharing
- Reduces parameters compared to fully connected networks
- Achieves state-of-the-art results in many vision tasks


## Why CNN is better than ANN for image processing

- **Weight Sharing**: CNNs share weights across all pixels in an image, which reduces the number of parameters needed to be learned. This makes training faster and more efficient.
- **Spatial Information**: CNNs take into account the spatial structure of an image, which is important for feature extraction. ANNs lose this information as they flatten the image data.
- **Automatic Feature Learning**: CNNs can learn features from raw image data, eliminating the need for manual feature engineering. ANNs require manual feature engineering, which can be time-consuming and may not result in the best features for the task at hand.
