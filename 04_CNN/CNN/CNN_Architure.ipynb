
## Complete CNN Architecture with Diagram

The complete CNN architecture is as follows:


Input → Convolutional Layer → Activation Function (ReLU) → Pooling Layer (Max Pooling) → Flatten Layer → Dense Layer (Softmax) → Output



### Complete Detail Example

Consider the following example:

* Input: 28x28 greyscale image of a handwritten digit (MNIST dataset)
* Convolutional layer: 32 filters of size 3x3, ReLU activation, Max Pooling with stride 2
* Flatten layer: convert 2D feature map to 1D feature vector
* Dense layer: 10 neurons with Softmax activation (output layer)
* Output: probability distribution over 10 digits (0-9)

The output of the convolutional layer is a 2D feature map of size 26x26x32. After max pooling, the feature map is reduced to 13x13x32. After flattening, the feature map is converted to a 1D feature vector of size 5408. Finally, the output layer produces a probability distribution over 10 digits.
