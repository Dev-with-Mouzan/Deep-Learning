## Convolution Layer in CNN

A **convolution layer** is the core building block of a Convolutional Neural Network (CNN).

### What it does
- Applies a set of learnable filters (kernels) to the input image or feature map.
- Computes local feature responses by sliding each filter over the input.
- Produces new feature maps that capture patterns like edges, textures, and shapes.

### How it works
1. A filter is a small matrix of weights, for example `3x3` or `5x5`.
2. The filter is moved across the input spatially.
3. At each position, the element-wise product between the filter and the corresponding input patch is computed.
4. The results are summed to produce one value in the output feature map.
5. This operation repeats for every position and every filter.

### Key properties
- **Local connectivity**: each output value depends only on a local region of the input.
- **Shared weights**: the same filter weights are used across the entire input.
- **Sparse interactions**: each output neuron connects to a small patch of the input.

### Main hyperparameters
- **Filter size**: often `3x3`, `5x5`, etc.
- **Number of filters**: controls the depth of the output volume.
- **Stride**: how many pixels the filter moves at each step.
- **Padding**: adding border pixels to control output size.
  - `valid` padding: no padding
  - `same` padding: output size remains the same as input size

### Output
- Produces a set of feature maps.
- Each feature map corresponds to one filter.
- The output volume size depends on input dimensions, filter size, stride, and padding.

### Why it is important
- Extracts spatial features automatically from raw input.
- Learns hierarchical representations:
  - early layers detect low-level features (edges, corners)
  - deeper layers detect high-level concepts (objects, parts)
- Reduces the number of parameters compared to fully connected layers.

### Example
For an input image of size `32x32x3` and `10` filters of size `3x3`, stride `1`, padding `same`, the output size is `32x32x10`.

### Keras Code Example

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

model = Sequential([
    Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu', padding='same'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()

# Example training call:
# model.fit(x_train, y_train, batch_size=64, epochs=10, validation_split=0.2)
```
