
# What is Padding in CNN?

In CNNs, we use a technique called padding to preserve the spatial size of the output. Padding is a process of adding zeros around the border of the image. The main purpose of padding is to maintain the spatial size of the output after applying convolutional and pooling operations.

There are two types of padding in CNNs:

1. **Valid Padding**: This type of padding does not add any zeros to the input. The output size is calculated as follows: output_size = (input_size - filter_size) + 1.

2. **Same Padding**: This type of padding adds zeros to the input such that the output size is the same as the input size. The output size is calculated as follows: output_size = input_size.



# What is Stride in CNN?

Stride is a hyperparameter in CNNs that determines how much the filter moves when applying the convolutional operation. A stride of 1 means that the filter moves 1 pixel at a time, while a stride of 2 means that the filter moves 2 pixels at a time.

The output size of a convolutional layer is calculated as follows: output_size = (input_size - filter_size + 2*padding) / stride + 1.

The main purpose of stride is to reduce the spatial size of the output. This can reduce the number of parameters and computations in the network, which can improve training speed and reduce overfitting.
