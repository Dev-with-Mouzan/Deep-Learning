```python
import numpy as np 
from keras.utils import image_dataset_from_directory
from keras.layers import RandomFlip, RandomRotation, RandomZoom,Dense,Dropout,GlobalAveragePooling2D,BatchNormalization,Rescaling
from keras.models import Sequential
from keras.applications import ResNet50
from keras.callbacks import EarlyStopping
```

```python
train_ds = image_dataset_from_directory(
    "animals",
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=(224, 224),
    batch_size=32
)
val_ds = image_dataset_from_directory(
    "animals",
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=(224, 224),
    batch_size=32
)
```

```python
DataAugmentation = Sequential([
    RandomFlip("horizontal"),
    RandomRotation(0.2),
    RandomZoom(0.2)
])
```

```python
Base_Model = ResNet50(weights="imagenet", include_top=False, input_shape=(224, 224, 3))
Base_Model.trainable = False
```

```python
model=Sequential([
    DataAugmentation,
    Rescaling(1./255),
    Base_Model,
    GlobalAveragePooling2D(),
    BatchNormalization(),
    Dense(512, activation="relu"),
    Dropout(0.5),
    Dense(256, activation="relu"),
    Dropout(0.5),
    Dense(1, activation="sigmoid")
])
```

```python
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
```

```python
model.fit(train_ds, validation_data=val_ds, epochs=10, callbacks=[EarlyStopping(patience=3, restore_best_weights=True)])
```

```python
loss,accuracy=model.evaluate(val_ds)
```
