# What is RNN




RNN is a type of Sequential model which is designed to work on sequence data like text, time series and speech etc. For example, RNN can be used to predict the next word in a sentence based on the previous words.
RNN Stand for Recurrent Neural Network. RNN is a type of Sequential model which is design to work on sequence data like text,time series and speach e.t.c



# Why do we need RNN?

We need RNN because traditional feedforward neural networks do not support sequence data. For example, if we want to predict the next word in a sentence based on the previous words, traditional feedforward neural networks do not support this type of data. RNN is designed to work on sequence data and it remembers the previous state of the data which makes it suitable for tasks like language modeling, speech recognition, etc.
