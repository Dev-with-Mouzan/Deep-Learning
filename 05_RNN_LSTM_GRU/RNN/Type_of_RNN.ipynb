# Types of RNN (Recurrent Neural Networks)

Recurrent Neural Networks (RNNs) can be classified into four main types based on how the input and output sequences are structured.

---

## 1. One to One

- **Input:** Single data point
- **Output:** Single prediction
- **Use Case:** Traditional neural network / classification

```
Input (x) ──► RNN ──► Output (y)
```

**Example:** Image classification where a single image is input and a single class label is output.

---

## 2. One to Many

- **Input:** Single data point
- **Output:** Sequence of predictions
- **Use Case:** Image captioning, music generation

```
       ┌──► Output (y1)
Input (x) ──► RNN ──► Output (y2)
       └──► Output (y3)
```

**Example:** Given an image, generate a sequence of words describing the image (image captioning).

---

## 3. Many to One

- **Input:** Sequence of data points
- **Output:** Single prediction
- **Use Case:** Sentiment analysis, spam detection

```
Input (x1) ─┐
Input (x2) ─┼──► RNN ──► Output (y)
Input (x3) ─┘
```

**Example:** Analyzing a movie review (sequence of words) to classify it as positive or negative sentiment.

---

## 4. Many to Many

- **Input:** Sequence of data points
- **Output:** Sequence of predictions
- **Use Case:** Machine translation, video frame labeling

```
Input (x1) ─┐         ┌──► Output (y1)
Input (x2) ─┼──► RNN ──┼──► Output (y2)
Input (x3) ─┘         └──► Output (y3)
```

**Example:** Translating an English sentence (sequence of words) to French (sequence of translated words).

---

## Summary Table

| Type | Input | Output | Common Applications |
|------|-------|--------|---------------------|
| One to One | Single | Single | Simple classification |
| One to Many | Single | Sequence | Image captioning, Music generation |
| Many to One | Sequence | Single | Sentiment analysis, Spam detection |
| Many to Many | Sequence | Sequence | Machine translation, Video labeling |
