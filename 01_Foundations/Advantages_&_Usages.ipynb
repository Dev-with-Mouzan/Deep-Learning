
# Usages of Ativation with Diagram
 Activation functions are used to introduce non-linearity into the output of a neuron. Without an activation function, the output of each neuron would be a linear function of the input, which would limit the power of the neural network to learn complex relationships between input and output.





```python

from keras.utils import plot_model
from keras.models import Sequential
from keras.layers import Dense

# create a simple neural network
model = Sequential()
model.add(Dense(64, activation='relu', input_dim=100))
model.add(Dense(32, activation='relu'))
model.add(Dense(10, activation='softmax'))

# plot the model
plot_model(model, to_file='neural_network_diagram.png', show_shapes=True, show_layer_names=True)

print('Neural network diagram saved to neural_network_diagram.png')

# the diagram shows the usage of different activation functions:
#   - the first layer uses the ReLU activation function
#   - the second layer uses the ReLU activation function
#   - the third layer uses the softmax activation function
```


# Advantages of Activation Function:

1.  Introduce non-linearity: Activation functions introduce non-linearity into the output of a neuron, which allows the neural network to learn and represent more complex relationships between the input and output.

2.  Handle vanishing/exploding gradients: Activation functions can help to handle vanishing or exploding gradients during backpropagation by introducing non-linearity and preventing the gradients from growing too large.

3.  Improve training stability: Activation functions can improve training stability by introducing non-linearity and preventing the gradients from growing too large.

4.  Allow for complex decision boundaries: Activation functions can allow for complex decision boundaries to be learned by the neural network.

5.  Allow for multi-class classification: Activation functions can allow for multi-class classification by mapping the output of the neuron to a probability distribution over the possible classes.

6.  Allow for autoencoders: Activation functions can allow for autoencoders to be used by mapping the output of the neuron to a lower dimensional representation of the input.

7.  Allow for generative models: Activation functions can allow for generative models to be used by mapping the output of the neuron to a probability distribution over the possible outputs.
