
# What is Activation Function?
Activation functions are used in neural networks to introduce non-linearity into the output of a neuron. Without an activation function, the output of each neuron would be a linear function of the input, which would limit the power of the neural network to learn complex relationships between input and output. Activation functions generally have an S-shaped curve, which maps the input to the output of the neuron. The most commonly used activation functions are sigmoid, tanh, ReLU (Rectified Linear Unit), and softmax.


# Type of Activation Function?
The choice of activation function depends on the type of problem being tackled. For example:

*   Sigmoid is often used for binary classification problems, as it maps the input to a value between 0 and 1.
*   Tanh is often used for classification problems with more than two classes, as it maps the input to a value between -1 and 1.
*   ReLU is often used for hidden layers, as it is computationally efficient and can be used to introduce non-linearity into the output of a neuron.
*   Softmax is often used for output layers of classification problems, as it maps the input to a probability distribution over the possible classes.



# Linear Activation Function

The linear activation function is the simplest of all the activation functions. It maps the input directly to the output without any transformation. The linear activation function is defined as:

f(x) = x

The linear activation function is rarely used as it does not introduce any non-linearity into the output of a neuron. However, it can be used in the output layer of a neural network when the problem is a regression problem.



# Sigmoid Activation Function

The sigmoid activation function is an S-shaped curve that maps the input to a value between 0 and 1. It is often used for binary classification problems, as it maps the input to a probability distribution over the two classes. The sigmoid activation function is defined as:

f(x) = 1 / (1 + exp(-x))

Where exp is the exponential function. The sigmoid activation function is differentiable, which makes it easy to use in backpropagation. However, it has a drawback of producing outputs that are not zero-centered, which can slow down the training process.


# Tanh Activation Function

The tanh activation function is also an S-shaped curve, but it maps the input to a value between -1 and 1. It is often used for classification problems with more than two classes, as it maps the input to a value that is zero-centered. The tanh activation function is defined as:

f(x) = 2 / (1 + exp(-2x)) - 1

Where exp is the exponential function. The tanh activation function is differentiable, which makes it easy to use in backpropagation. It also produces outputs that are zero-centered, which can speed up the training process.


# ReLU Activation Function

The ReLU (Rectified Linear Unit) activation function is a simple and computationally efficient function that maps the input to the output of a neuron. The ReLU activation function is defined as:

f(x) = max(0,x)

The ReLU activation function is differentiable, which makes it easy to use in backpropagation. It is also computationally efficient, as it does not require the computation of an exponential function. The ReLU activation function is often used in hidden layers, as it can introduce non-linearity into the output of a neuron without significantly slowing down the training process.


# Leaky ReLU Activation Function

The Leaky ReLU (Rectified Linear Unit) activation function is a variant of the ReLU function that addresses the "dying ReLU" problem, where neurons can become inactive and stop learning. Instead of outputting zero for negative inputs, Leaky ReLU allows a small, positive gradient for negative values. The Leaky ReLU activation function is defined as:

f(x) = max(αx, x)

Where α is a small constant (typically 0.01). This allows the function to maintain a small gradient for negative inputs, preventing neurons from dying and improving training stability. Leaky ReLU is differentiable and computationally efficient, making it suitable for hidden layers in neural networks.


# PReLU Activation Function

The PReLU (Parametric ReLU) activation function is a variant of the Leaky ReLU function that addresses the problem of having a fixed slope for negative inputs. Instead of using a fixed slope, PReLU allows the model to learn an optimal slope for negative inputs. The PReLU activation function is defined as:

f(x) = max(αx, x)

Where α is a learnable parameter. This allows the model to learn an optimal slope for negative inputs, improving training stability and performance. PReLU is differentiable and computationally efficient, making it suitable for hidden layers in neural networks.



# Swish Activation Function

The Swish activation function is a self-gated activation function that adapts to the input distribution during training. It is defined as:

f(x) = x * g(x)

Where g(x) is a learnable function that maps the input to a value between 0 and 1. The Swish activation function is differentiable and computationally efficient, making it suitable for hidden layers in neural networks. It has been shown to outperform the ReLU and Leaky ReLU activation functions in some cases.
