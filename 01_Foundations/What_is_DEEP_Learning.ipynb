# Why do we use Deep Learning?

- Traditional ML algorithms are limited to simple decision boundaries (linear, quadratic, etc.).
- DL can learn complex decision boundaries by stacking multiple layers of neurons.
- DL can automatically extract relevant features from raw data, eliminating the need for manual feature engineering.
- DL has been shown to achieve state-of-the-art results in various applications, such as image recognition, speech recognition, and natural language processing.

# Deep learning (DL) vs machine learning (ML)

ML is broad: algorithms that learn patterns from data (linear regression, decision trees, SVM, etc.).
DL is a subset of ML: uses deep neural networks (many layers) to automatically learn feature representations.
ML often needs manual feature engineering; DL learns features end-to-end from raw data.
DL excels on large-scale, high-dimensional data (images, audio, text), ML is often better on smaller datasets and simpler problems.
DL models are generally more computationally intensive and less interpretable than many classic ML models.
