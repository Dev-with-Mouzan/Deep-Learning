## Gradient Descent Variants

**Gradient Descent** is an optimization algorithm used to minimize the loss function in machine learning models. It works by iteratively adjusting the model parameters in the direction of the negative gradient of the loss function.

### Basic Gradient Descent
- **Batch Gradient Descent**: Uses the entire dataset to compute the gradient in each iteration.
  - Pros: Stable convergence, accurate gradient estimate
  - Cons: Slow for large datasets, memory intensive
- **Stochastic Gradient Descent (SGD)**: Uses one sample at a time to compute the gradient.
  - Pros: Fast, can escape local minima, works with large datasets
  - Cons: Noisy updates, may not converge smoothly

### Advanced Variants

#### 1. Mini-Batch Gradient Descent
- Uses a small batch of samples (e.g., 32-256) to compute the gradient.
- Balances between batch GD and SGD.
- Most commonly used in practice.

#### 2. Momentum
- Accelerates SGD by adding a fraction of the previous update to the current update.
- Helps overcome oscillations and local minima.
- Formula: `v = βv + (1-β)∇θJ(θ)`
- Then: `θ = θ - αv`

#### 3. Nesterov Accelerated Gradient (NAG)
- Looks ahead before taking the step.
- Computes gradient at the "lookahead" position.
- Generally converges faster than momentum.

#### 4. Adagrad (Adaptive Gradient)
- Adapts the learning rate for each parameter based on historical gradients.
- Parameters with frequent updates get smaller learning rates.
- Formula: `θ = θ - (α / √(G + ε)) * ∇θJ(θ)`
- Where G is the sum of squared gradients.

#### 5. RMSProp (Root Mean Square Propagation)
- Addresses Adagrad's rapidly decreasing learning rates.
- Uses exponentially weighted moving average of squared gradients.
- Formula: `E[g²] = βE[g²] + (1-β)g²`
- `θ = θ - (α / √(E[g²] + ε)) * g`

#### 6. Adam (Adaptive Moment Estimation)
- Combines momentum and RMSProp.
- Maintains both first moment (mean) and second moment (variance) of gradients.
- Most popular optimizer in deep learning.
- Formulas:
  - `m = β₁m + (1-β₁)g` (first moment)
  - `v = β₂v + (1-β₂)g²` (second moment)
  - Bias correction: `m̂ = m/(1-β₁ᵗ)`, `v̂ = v/(1-β₂ᵗ)`
  - `θ = θ - (α * m̂) / (√v̂ + ε)`

#### 7. AdaMax
- Variant of Adam using infinity norm instead of L2 norm.
- More stable for some problems.

#### 8. Nadam (Nesterov + Adam)
- Combines Nesterov momentum with Adam.

### Choosing the Right Optimizer
- **SGD with momentum**: Good baseline, works well with proper tuning.
- **Adam**: Default choice for most deep learning tasks.
- **RMSProp**: Good for RNNs and when dealing with sparse gradients.
- **Adagrad**: Useful for sparse data, but learning rate may decay too fast.

### Hyperparameters
- **Learning Rate (α)**: Step size. Too high: may not converge. Too low: slow convergence.
- **Momentum (β)**: Usually 0.9 for momentum, 0.999 for Adam's β₂.
- **Epsilon (ε)**: Small constant to prevent division by zero (e.g., 1e-8).

### Tips for Better Convergence
- Use learning rate schedules (decay over time).
- Implement early stopping based on validation loss.
- Normalize input features.
