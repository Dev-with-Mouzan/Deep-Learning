
# Batch Normalization
Batch normalization is a technique used to normalize input data in a neural network. It is applied to each mini-batch of data separately, normalizing the data to have a mean of 0 and standard deviation of 1. This is done by subtracting the mean of the mini-batch from each element and dividing by the standard deviation of the mini-batch.

The batch normalization process can be formulated as follows:

X = X - mu / sqrt(var + eps)

where mu is the mean of the mini-batch, var is the variance of the mini-batch, and eps is a small value added for numerical stability.

The batch normalization process is typically done after each convolutional or fully connected layer in a neural network. It is applied to the input of the layer before the layer's weights are applied.

The main purpose of batch normalization is to reduce the problem of internal covariate shift. Internal covariate shift occurs when the distribution of the inputs to a layer changes during training, which can cause the layer's weights to change in a way that is not desirable. Batch normalization helps to reduce this problem by normalizing the input data to have a mean of 0 and standard deviation of 1, which helps to stabilize the training process.

Batch normalization also has the effect of regularizing the model, which can help to reduce overfitting. Regularization is a technique used to reduce overfitting by adding a penalty term to the loss function. The penalty term is proportional to the magnitude of the model's weights, and is used to discourage the model from having large weights. Batch normalization helps to reduce overfitting by normalizing the input data to have a mean of 0 and standard deviation of 1, which helps to reduce the magnitude of the model's weights.


```python

from tensorflow.keras.layers import BatchNormalization

class BatchNormKeras:
    def __init__(self, **kwargs):
        self.bn = BatchNormalization(**kwargs)

    def __call__(self, x):
        return self.bn(x)
```
