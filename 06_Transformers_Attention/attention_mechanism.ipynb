# Attention Mechanism

The attention mechanism is a technique that helps neural networks focus on the most relevant parts of the input when processing or generating output. Instead of treating all inputs equally, attention assigns different importance weights to different parts of the input.

## Problem it solves

- **Fixed context bottleneck**: In standard Seq2Seq models, the entire input must be compressed into a single context vector, which loses information for long sequences.
- **Long-range dependencies**: Traditional models struggle to connect distant input and output elements.

## How attention works

1. **Query, Key, Value**: The decoder (query) searches through encoder outputs (keys and values).
2. **Compute similarity scores**: Calculate how relevant each input token is to the current output token.
3. **Create weights**: Normalize scores into attention weights that sum to 1.
4. **Weighted sum**: Combine input values using these weights to create a context-aware representation.

## Key concepts

- **Query (Q)**: What the decoder is looking for.
- **Key (K)**: What each encoder output represents; used to measure relevance.
- **Value (V)**: The actual encoder output information to be combined.
- **Attention weights**: Normalized scores showing how much attention each input deserves.

## Types of attention

- **Additive attention (Bahdanau)**: Uses a neural network to compute similarity scores.
- **Multiplicative attention (Dot-product)**: Computes similarity as a dot product of queries and keys.
- **Multi-head attention**: Multiple attention operations in parallel to capture different relationships.

## Benefits

- Improves translation quality by aligning source and target words.
- Makes the model interpretable by showing which input parts influenced each output.
- Handles long sequences better by focusing selectively rather than compressing everything.
- Enables better learning of dependencies between distant elements.

## Summary

Attention mechanisms allow neural networks to dynamically focus on the most relevant input elements when generating output, solving the context bottleneck problem and improving performance on sequence-to-sequence tasks.
