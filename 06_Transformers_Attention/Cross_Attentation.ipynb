# Cross-attention
Cross-attention is the attention mechanism used in Transformer decoder layers to attend from one sequence to another.

### What it is
- **Source**: Queries come from the decoder hidden states.
- **Target**: Keys and values come from the encoder hidden states.
- This lets the decoder use encoded information from the input sequence while generating each output token.

### How it works
1. Compute decoder queries \(Q\) from the decoder input.
2. Compute encoder keys \(K\) and values \(V\) from the encoder output.
3. Compute attention scores: \( \text{scores} = Q K^T / \sqrt{d_k} \).
4. Apply softmax to get attention weights.
5. Multiply weights by \(V\) to produce the context vector.

### Why it matters
- It connects the encoder and decoder in encoder-decoder models.
- It lets the decoder “look at” the encoded input when generating outputs.
- Common use case: machine translation, summarization, and any seq2seq Transformer.

### Difference from self-attention
- **Self-attention**: \(Q, K, V\) all come from the same sequence.
- **Cross-attention**: \(Q\) comes from one sequence and \(K, V\) come from another.
