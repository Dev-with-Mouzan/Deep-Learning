# what is tranformer? 
The Transformer is a groundbreaking neural network architecture introduced in the 2017 paper "Attention is All You Need" by Ashish Vaswani, Noam Shazeer, Niki Parmar, Jakob Uszkoreit, Llion Jones, Aidan N. Gomez, Łukasz Kaiser, and Illia Polosukhin. It revolutionized natural language processing (NLP) and other sequence-based tasks by relying entirely on attention mechanisms, eliminating the need for recurrent neural networks (RNNs) like LSTMs or GRUs.

### Key Concepts:
- **Core Idea**: Transformers process sequences (e.g., text) using self-attention, allowing the model to weigh the importance of different parts of the input when generating outputs. This enables parallel processing (unlike RNNs) and better handling of long-range dependencies.
- **Architecture**:
  - **Encoder**: Processes the input sequence into a set of representations. Consists of multiple layers, each with multi-head self-attention and feed-forward networks.
  - **Decoder**: Generates the output sequence, using cross-attention to attend to encoder outputs and self-attention for the target sequence.
  - **Attention Mechanisms**:
    - **Self-Attention**: Computes relationships within the same sequence (e.g., how words relate to each other in a sentence).
    - **Multi-Head Attention**: Splits attention into multiple "heads" for richer representations.
    - **Scaled Dot-Product Attention**: The core attention formula: \( \text{Attention}(Q, K, V) = \softmax\left(\frac{QK^T}{\sqrt{d_k}}\right)V \), where Q, K, V are query, key, and value matrices.
  - **Positional Encoding**: Adds information about token positions since Transformers don't have inherent sequence order.
  - **Feed-Forward Networks**: Applied per token after attention.
- **Advantages**:
  - Highly parallelizable (faster training on GPUs).
  - Scales better to long sequences.
  - Foundation for models like BERT, GPT, T5, and many others.
- **Variants**: Encoder-only (e.g., BERT for understanding), Decoder-only (e.g., GPT for generation), Encoder-Decoder (e.g., original Transformer for translation).
