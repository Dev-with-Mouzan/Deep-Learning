Luong attention (also known as multiplicative attention or global attention) is a neural attention mechanism introduced in the 2015 paper "Effective Approaches to Attention-based Neural Machine Translation" by Minh-Thang Luong, Hieu Pham, and Christopher D. Manning. It's a variant of the Bahdanau attention mechanism, designed to be more efficient and effective for sequence-to-sequence (seq2seq) models, particularly in machine translation.

### Key Concepts:
- **Purpose**: Like Bahdanau attention, it allows the decoder to focus on relevant parts of the encoder's output when generating each target token, enabling better handling of long sequences and improving translation quality.
- **How it works**: At each decoder timestep, it computes attention scores by taking the dot product (or a more complex function) between the decoder's hidden state and each encoder hidden state. These scores are normalized into attention weights, which create a context vector as a weighted sum of encoder states.
- **Types of Luong Attention**:
  - **Dot-product attention**: Simplest form, score = \( s_i^T h_j \) (where \( s_i \) is decoder state, \( h_j \) is encoder state).
  - **General attention**: Adds a weight matrix, score = \( s_i^T W_a h_j \).
  - **Concat attention**: Similar to Bahdanau, score = \( v_a^T \tanh(W_a [s_i; h_j]) \).
- **Advantages over Bahdanau**: Often faster to compute (especially dot-product version) and can perform comparably or better on certain tasks. It also supports different alignment functions.
- **Global vs. Local**: Luong proposed both global (attends to all encoder states) and local (attends to a window around predicted position) variants.

### Comparison to Bahdanau:
- Bahdanau uses additive scoring (concat + tanh + linear), while Luong uses multiplicative (dot-product or matrix multiplication).
- Luong attention is applied after the decoder RNN output, whereas Bahdanau applies it before.

This mechanism is widely used in modern seq2seq architectures.
