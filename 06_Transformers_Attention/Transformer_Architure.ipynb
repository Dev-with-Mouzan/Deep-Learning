## Transformer Architecture in “Attention Is All You Need”

The Transformer is an encoder-decoder model built entirely on attention mechanisms, removing recurrence and convolution.

### Encoder
- Consists of a stack of identical layers.
- Each layer has:
  1. **Multi-Head Self-Attention**
     - Each position attends to all positions in the input.
     - Multiple heads capture different relationships.
  2. **Position-wise Feed-Forward Network**
     - Two linear layers with ReLU.
  3. **Residual Connections + Layer Normalization**
     - Add & normalize after attention and after feed-forward.

### Decoder
- Also a stack of identical layers.
- Each layer has:
  1. **Masked Multi-Head Self-Attention**
     - Prevents attending to future positions during generation.
  2. **Multi-Head Cross-Attention**
     - Queries come from the decoder, keys/values come from the encoder output.
  3. **Position-wise Feed-Forward Network**
  4. **Residual Connections + Layer Normalization**

### Attention
- Core formula:  
  `Attention(Q, K, V) = softmax(Q K^T / sqrt(d_k)) V`
- **Multi-head** means this is done in parallel with different learned projections, then concatenated.

### Positional Encoding
- Since the model processes tokens in parallel, positional encodings are added to token embeddings to give the model sequence order.
- The original paper uses sinusoidal positional encoding:
  - `sin(pos / 10000^(2i/d))` for even dimensions
  - `cos(pos / 10000^(2i/d))` for odd dimensions

### Why it matters
- Enables parallel training over full sequences.
- Handles long-range dependencies better than RNNs.
- Became the foundation for BERT, GPT, T5, and many modern NLP models.

### Summary
The Transformer architecture in the paper is:
- Encoder-decoder
- Attention-only
- Uses multi-head attention
- Uses positional encodings
- Uses residual + layer norm + feed-forward blocks
