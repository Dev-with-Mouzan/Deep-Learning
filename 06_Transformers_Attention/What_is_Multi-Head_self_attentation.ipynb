

### Intuition
- **Why Multi-Head?** Single self-attention computes one set of attention weights per position, which might miss diverse relationships (e.g., syntactic vs. semantic dependencies in a sentence). Multi-head attention runs multiple parallel "heads" of self-attention, each learning different aspects of the relationships.
- **Analogy**: Like having multiple experts (heads) analyzing the same data from different angles, then combining their insights.

### How It Works
Assume an input sequence \( X = [x_1, x_2, \dots, x_n] \), where each \( x_i \in \mathbb{R}^d \) (d is the model dimension, e.g., 512).

1. **Splitting into Heads**:
   - The model dimension \( d \) is divided into \( h \) heads (e.g., 8 heads for \( d=512 \), so each head has dimension \( d_h = d / h = 64 \)).
   - For each head \( i \), project \( X \) into separate Query (\( Q_i \)), Key (\( K_i \)), and Value (\( V_i \)) matrices using learned weights \( W_i^Q, W_i^K, W_i^V \in \mathbb{R}^{d \times d_h} \).
     \[
     Q_i = X W_i^Q, \quad K_i = X W_i^K, \quad V_i = X W_i^V
     \]
   - Each head operates independently on its own \( Q_i, K_i, V_i \).

2. **Self-Attention per Head**:
   - For each head \( i \), compute scaled dot-product attention:
     \[
     \text{head}_i = \text{Attention}(Q_i, K_i, V_i) = \softmax\left(\frac{Q_i K_i^T}{\sqrt{d_h}}\right) V_i
     \]
   - Output per head: \( \text{head}_i \in \mathbb{R}^{n \times d_h} \).

3. **Concatenation and Projection**:
   - Concatenate all head outputs: \( \text{Concat}(\text{head}_1, \dots, \text{head}_h) \in \mathbb{R}^{n \times d} \).
   - Apply a final linear projection \( W^O \in \mathbb{R}^{d \times d} \) to mix information across heads:
     \[
     \text{MultiHead}(Q, K, V) = \text{Concat}(\text{head}_1, \dots, \text{head}_h) W^O
     \]
   - Final output: \( \mathbb{R}^{n \times d} \), same shape as input.

### Key Benefits
- **Diverse Representations**: Different heads can capture different types of dependencies (e.g., one head for subject-verb agreement, another for long-distance pronouns).
- **Improved Performance**: Empirically, multi-head attention outperforms single-head versions on tasks like translation and language modeling.
- **Parallel Computation**: All heads can be computed in parallel, maintaining efficiency.



### In Transformers
- Used in both encoder (self-attention on input) and decoder (self-attention on target + cross-attention to encoder).
- Often followed by feed-forward networks and residual connections.
