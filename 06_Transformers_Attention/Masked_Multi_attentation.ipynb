# Masked multi-head attention
Masked multi-head attention is the decoder self-attention mechanism used in Transformer models to prevent each position from seeing future tokens during training.

### What it means
- **Masked**: A mask is applied so that when the model processes position `i`, it can only attend to positions `<= i`. This preserves the autoregressive property needed for generation.
- **Multi-head**: The attention is computed in parallel across multiple heads, letting the model capture different types of relationships simultaneously.
- **Self-attention**: Queries, keys, and values all come from the same input sequence (the decoder input).

### Why it is used
- In the decoder, you must not let the model “peek ahead” at later words when predicting the next word.
- Without masking, the decoder could use future target tokens and learn an invalid mapping.

### How it works
1. Compute `Q`, `K`, `V` from the decoder input.
2. Compute attention scores: `scores = Q K^T / sqrt(d_k)`.
3. Apply a causal mask so that `scores[i, j] = -inf` for `j > i`.
4. Softmax the masked scores to get attention weights.
5. Multiply by `V` and combine heads.

### Common use case
- In language generation (`GPT`, Transformer decoder), this is used in every decoder self-attention layer.
- It ensures the model generates text one token at a time.
