# what is Bahdanau attention ?
Bahdanau attention (also known as additive attention) is a neural attention mechanism introduced in the 2014 paper "Neural Machine Translation by Jointly Learning to Align and Translate" by Dzmitry Bahdanau, Kyunghyun Cho, and Yoshua Bengio. It's a foundational technique in sequence-to-sequence (seq2seq) models, particularly for machine translation tasks.

# Key Concepts:
<h4>Purpose:</h4>
 Allows the model to focus on relevant parts of the input sequence when generating each output element, rather than treating the entire input as a fixed-length vector.

<h4>How it works:</h4>
 For each decoder timestep, it computes attention scores by comparing the decoder's hidden state with all encoder hidden states using a learned alignment function (typically a feedforward network). These scores are normalized into weights, which are used to create a weighted sum of encoder 
 states as context for the decoder.

# Formula:
 The attention score for encoder state ( h_j ) and decoder state ( s_i ) is computed as:
[
score(s_i, h_j) = v_a^T \tanh(W_a s_i + U_a h_j)
]
where ( v_a ), ( W_a ), and ( U_a ) are learnable parameters.

# Advantages:
 Improves translation quality by allowing soft alignment between source and target words, handling variable-length sequences better than fixed-length encodings.
This mechanism paved the way for modern attention-based architectures like Transformers.
