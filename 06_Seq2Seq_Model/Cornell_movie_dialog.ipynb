```python
def movie_line(line):
    lines = {}
    with open(line, encoding='utf-8', errors='ignore') as f:
        for line in f:
            parts = line.split(" +++$+++ ")
            lines[parts[0]] = parts[-1].strip()
    return lines
```

```python
def load_conversations(line):
    convs = []
    with open(line, encoding='utf-8', errors='ignore') as f:
        for line in f:
            parts = line.split(" +++$+++ ")
            ids = eval(parts[-1])  # list of line IDs
            convs.append(ids)
    return convs
```

```python
lines = movie_line("movie_lines.txt")
conversations = load_conversations("movie_conversations.txt")
```

```python
lines
```

```python
question=[]
answere=[]

for conv in conversations:
    for i in range(len(conv) - 1):
        question.append(lines[conv[i]])
        answere.append(lines[conv[i + 1]])
```

```python
clean_q = []
clean_a = []

for q, a in zip(question, answere):
    if 2 < len(q.split()) < 15 and 2 < len(a.split()) < 15:
        clean_q.append(q)
        clean_a.append(a)
```

```python
import random

combined = list(zip(clean_q, clean_a))
random.shuffle(combined)

clean_q, clean_a = zip(*combined)
```

```python
question = list(clean_q[:1000])
answere = list(clean_a[:1000])
```

```python
import re

def clean_text(txt):
  txt=txt.lower()
  txt=re.sub(r"[^a-zA-Z?.!,]+", " ", txt)
  txt=re.sub(r"\s+", " ", txt).strip()
  return txt
```

```python

question = [clean_text(q) for q in question]
answere = [clean_text(a) for a in answere]
```

```python
answere = ["<start> " + ans + " <end>" for ans in answere]
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python

tokenizer = Tokenizer()
tokenizer.fit_on_texts(question + answere)

q_seq = tokenizer.texts_to_sequences(question)
a_seq = tokenizer.texts_to_sequences(answere)

q_seq = pad_sequences(q_seq, maxlen=20, padding='post')
a_seq = pad_sequences(a_seq, maxlen=20, padding='post')
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    q_seq, a_seq, test_size=0.1
)
```

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Embedding, Dense

vocab_size = len(tokenizer.word_index) + 1

# Encoder
encoder_inputs = Input(shape=(None,))
enc_emb = Embedding(vocab_size, 256)(encoder_inputs)
encoder_lstm = LSTM(256, return_state=True)
_, state_h, state_c = encoder_lstm(enc_emb)

# Decoder
decoder_inputs = Input(shape=(None,))
dec_emb = Embedding(vocab_size, 256)(decoder_inputs)
decoder_lstm = LSTM(256, return_sequences=True, return_state=True)
decoder_outputs, _, _ = decoder_lstm(dec_emb, initial_state=[state_h, state_c])

decoder_dense = Dense(vocab_size, activation='softmax')
output = decoder_dense(decoder_outputs)

model = Model([encoder_inputs, decoder_inputs], output)
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy',metrics=["accuracy"])
```

```python
from keras.callbacks import EarlyStopping
model.fit(
    [X_train, y_train],
    y_train,
    batch_size=64,
    epochs=100,
    validation_data=([X_test, y_test], y_test),callbacks=[EarlyStopping(patience=3)]
)
```

```python
loss,accuracy=model.evaluate([X_test,y_test],y_test)
```

```python
print("loss: ",loss)
print("accuracy:",accuracy)
```
