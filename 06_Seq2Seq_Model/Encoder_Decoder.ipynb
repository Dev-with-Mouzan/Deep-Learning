# Encoder-Decoder Architecture

The encoder-decoder architecture is a neural network design used for sequence-to-sequence tasks. It consists of two main parts:

## Encoder

- The encoder reads the input sequence one step at a time.
- It converts the input into a hidden representation, often called the context vector or encoder states.
- In RNN-based models, the encoder is typically built with LSTM or GRU cells.
- The final hidden state summarizes the input sequence information.

## Decoder

- The decoder generates the output sequence one token at a time.
- It starts from the encoder's final state and uses that information to produce the first output.
- At each step, the decoder also receives the previously generated token as input.
- The decoder predicts the next token until it generates an end-of-sequence symbol.

## How they work together

1. The encoder processes the entire input sequence and produces a representation of its meaning.
2. The decoder takes that representation and generates a corresponding output sequence.

## Why it is useful

- It can map variable-length input sequences to variable-length output sequences.
- It supports tasks such as translation, summarization, and dialogue generation.
- Attention mechanisms can be added to allow the decoder to focus on relevant parts of the input rather than a single fixed vector.

## Summary

The encoder-decoder architecture separates encoding and decoding into two networks, making it easier to learn complex sequence transformations from input to output.
